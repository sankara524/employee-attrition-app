# Step 1: Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    ConfusionMatrixDisplay
)

print("Libraries Imported Successfully")
```

# Step 2: Load the Dataset

```python
df = pd.read_csv('/content/Employee Attrition.csv')
```

```python
df.head()
```

```python
df.tail()
```

```python
print("Dataset Verification")
print("=" * 60)

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nAll Dataset Columns:")
print(df.columns.tolist())
```

# Step 3: Feature and Target Separation

```python
# Remove Employee ID because it is an identifier, not a

X = df.drop(columns=['Employee ID', 'Attrition'])

# Target variable
y = df['Attrition']

print("Feature and Target Separation")
print("=" * 60)

print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nFeature Columns:")
print(X.columns.tolist())

print("\nTarget Values:")
print(y.unique())
```

# Step 4: Target Distribution

```python
print("Attrition Value Counts")
print("=" * 60)

print(y.value_counts())

print("\nAttrition Percentage")
print("=" * 60)

print(
    y.value_counts(normalize=True).mul(100).round(2)
)
```

# Step 5: Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)

print("Train-Test Split Completed")
print("=" * 60)

print("Training Feature Shape:", X_train.shape)
print("Testing Feature Shape :", X_test.shape)

print("Training Target Shape :", y_train.shape)
print("Testing Target Shape  :", y_test.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts())

print("\nTesting Target Distribution:")
print(y_test.value_counts())
```

# Step 6: Feature Type Identification

```python
numerical_features = X.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

categorical_features = X.select_dtypes(
    include=['object', 'category']
).columns.tolist()

print("Feature Type Identification")
print("=" * 60)

print("\nNumerical Features:")
print(numerical_features)

print("\nNumber of Numerical Features:", len(numerical_features))

print("\nCategorical Features:")
print(categorical_features)

print("\nNumber of Categorical Features:", len(categorical_features))
```

# Step 7: Preprocessing Pipeline

```python
numerical_transformer = Pipeline(
    steps=[
        ('scaler', StandardScaler())
    ]
)

categorical_transformer = Pipeline(
    steps=[
        ('onehot', OneHotEncoder(
            handle_unknown='ignore',
            sparse_output=False
        ))
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)

print("SVM Preprocessing Pipeline Created Successfully")
print("=" * 60)
print("Numerical Features:", len(numerical_features))
print("Categorical Features:", len(categorical_features))
```

# Step 8: Baseline SVM Pipeline

```python
svm_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('svm', SVC(
            kernel='rbf',
            probability=True,
            random_state=42
        ))
    ]
)

print("Baseline SVM Pipeline Created Successfully")
print("=" * 60)
print("Model: Support Vector Machine (SVM)")
print("Kernel: RBF")
print("Probability: True")
print("Random State: 42")
```

# Step 9: Train Baseline SVM

```python
svm_pipeline.fit(X_train, y_train)

print("Baseline SVM Model Trained Successfully")
print("=" * 60)
```

# Step 10: Baseline Predictions

```python
y_pred_svm = svm_pipeline.predict(X_test)

print("Baseline SVM Predictions Generated Successfully")
print("=" * 60)

print("Number of Predictions:", len(y_pred_svm))
print("First 20 Predictions:")
print(y_pred_svm[:20])
```

# Step 11: Baseline SVM Performance

```python
accuracy_svm = accuracy_score(y_test, y_pred_svm)
precision_svm = precision_score(y_test, y_pred_svm, pos_label='Left')
recall_svm = recall_score(y_test, y_pred_svm, pos_label='Left')
f1_svm = f1_score(y_test, y_pred_svm, pos_label='Left')

print("Baseline SVM Performance")
print("=" * 50)

print(f"Accuracy : {accuracy_svm:.4f}")
print(f"Precision: {precision_svm:.4f}")
print(f"Recall   : {recall_svm:.4f}")
print(f"F1 Score : {f1_svm:.4f}")
```

# Step 12: Confusion Matrix

```python
cm_svm = confusion_matrix(
    y_test,
    y_pred_svm,
    labels=['Stayed', 'Left']
)

print("Baseline SVM Confusion Matrix")
print("=" * 50)
print(cm_svm)

print("\nTotal samples in confusion matrix:", cm_svm.sum())
```

# Step 13: Classification Report

```python
print("Classification Report - Baseline SVM")
print("=" * 60)

print(
    classification_report(
        y_test,
        y_pred_svm,
        labels=['Stayed', 'Left'],
        digits=4
    )
)
```

# Step 14: Baseline ROC-AUC

```python
y_prob_svm = svm_pipeline.predict_proba(X_test)[:, 1]

roc_auc_svm = roc_auc_score(
    y_test,
    y_prob_svm
)

print("Baseline SVM ROC-AUC")
print("=" * 50)
print(f"ROC-AUC : {roc_auc_svm:.4f}")
```

# Step 15: Save Final SVM Model

```python
import joblib

svm_model_filename = "Employee_Attrition_SVM_Model.pkl"

joblib.dump(
    svm_pipeline,
    svm_model_filename
)

print("SVM Model Saved Successfully")
print("=" * 60)
print("File Name:", svm_model_filename)
```

# Step 16: Load Saved Model

```python
import joblib

svm_model_filename = "Employee_Attrition_SVM_Model.pkl"

loaded_svm_model = joblib.load(
    svm_model_filename
)

print("SVM Model Loaded Successfully")
print("=" * 60)
print("Loaded Model:", svm_model_filename)
```

# Step 17: Loaded Model Verification

```python
y_pred_loaded_svm = loaded_svm_model.predict(X_test)

print("Loaded SVM Model Verification")
print("=" * 60)

print("Number of Predictions:", len(y_pred_loaded_svm))

print("\nFirst 20 Predictions:")
print(y_pred_loaded_svm[:20])
```

# Step 18: Final Model Verification

```python
same_predictions_svm = np.array_equal(
    y_pred_svm,
    y_pred_loaded_svm
)

print("SVM Prediction Comparison")
print("=" * 60)

if same_predictions_svm:
    print("Original and Loaded Model Predictions: SAME")
    print("Model Verification: PASSED")
else:
    print("Original and Loaded Model Predictions: DIFFERENT")
    print("Model Verification: FAILED")
```
